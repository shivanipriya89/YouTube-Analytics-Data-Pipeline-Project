Q3) Data Cleaning
Clean the datasets using Python.

Tasks:
- Remove duplicates
- Convert dates to datetime
- Convert numeric columns to appropriate data types
- Handle missing values
- Standardize column names


```python
#3
print(df1.duplicated().sum()) # Total Sum of duplicate values of videos dataframe
print(df1.drop_duplicates(inplace=True)) # Dropping the duplicates of videos dataframe
df1["publish_date"]=pd.to_datetime(df1["publish_date"])
df1["video_title"]=df1["video_title"].astype("string")
df1["category"]=df1["category"].astype("string")
print(df1.dtypes) # Datatypes of columns of videos dataframe
print(df1.isnull().sum()) # Total Sum of Missing Values of videos dataframe
print(df1.dropna(inplace=True)) # Dropping the Null Values of videos dataframe


print(df2.duplicated().sum()) # Total Sum of duplicate values of channels dataframe
print(df2.drop_duplicates(inplace=True)) # Dropping the duplicates of channels dataframe
print(df2.isnull().sum()) # Total Sum of Missing Values of channels dataframe
print(df2.dropna(inplace=True)) # Dropping the Null Values of channels dataframe
df2["channel_name"]=df2["channel_name"].astype("string")
df2["channel_category"]=df2["channel_category"].astype("string")
df2["country"]=df2["country"].astype("string")
print(df2.dtypes) # Datatypes of columns


print(df3.duplicated().sum()) # Total Sum of duplicate values of video_metrics dataframe
print(df3.drop_duplicates(inplace=True)) # Dropping the duplicates of videos_metrics dataframe
print(df3.isnull().sum()) # Total Sum of Missing Values of Video metrics dataframe
print(df3.dropna(inplace=True)) # Dropping the Null Values of video metrics dataframe
df3["metric_date"]=pd.to_datetime(df3["metric_date"])
print(df3.dtypes) # Datatypes of Video metrics dataframe


print(df4.duplicated().sum()) # Total Sum of duplicate values of Traffic_Sources dataframe
print(df4.drop_duplicates(inplace=True)) # Dropping the duplicates of Traffic_Sources dataframe
print(df4.isnull().sum()) # Total Sum of Missing Values of Traffic_Sources dataframe
print(df4.dropna(inplace=True)) # Dropping the Null Values of Traffic_Sources dataframe
df4["traffic_date"]=pd.to_datetime(df4["traffic_date"])
df4["traffic_source"]=df4["traffic_source"].astype("string")
print(df4.dtypes)

# Dropping the Null Values and duplicates of all 4 dataframes
print(df1.dropna(inplace=True))
print(df1.drop_duplicates(inplace=True))

print(df2.dropna(inplace=True))
print(df2.drop_duplicates(inplace=True))


print(df3.dropna(inplace=True))
print(df3.drop_duplicates(inplace=True))


print(df4.dropna(inplace=True))
print(df4.drop_duplicates(inplace=True))

# Saving the Cleaned dataset into cleaned data
df1.to_csv("cleaned_videos.csv",index=False)
df2.to_csv("cleaned_channels.csv",index=False)
df3.to_csv("cleaned_video_metrics.csv",index=False)
df4.to_csv("cleaned_traffics.csv",index=False)


```

Q4) Create an ETL Pipeline
Create a Python ETL pipeline:

Extract → Transform → Load

Extract the CSV files, transform the data, and load the cleaned data into MySQL staging tables.


```python
#4
import pandas as pd
def extract():
  df1=pd.read_csv("videos.csv")
  df2=pd.read_csv("channels.csv")
  df3=pd.read_csv("video_metrics.csv")
  df4=pd.read_csv("traffic_sources.csv")
  print(df1) # Videos dataframe
  print(df2) # Dataframe of channels
  print(df3) # Video metrics dataframe
  print(df4) # Traffic Sources dataframe
  return df1,df2,df3,df4
extract()
```

```python
def transform(df1,df2,df3,df4):
  print(df1.dtypes) # Datatypes of Videos dataframe
  print(df2.dtypes) # Datatypes of Channel dataframe
  print(df3.dtypes) # Datatypes of Video-Metrics dataframe
  print(df4.dtypes) # Datatypes of Traffic Sources dataframe
  # Dropping the Null Values and duplicates of all dataframe
  print(df1.drop_duplicates(inplace=True))
  print(df2.drop_duplicates(inplace=True))
  print(df3.drop_duplicates(inplace=True))
  print(df4.drop_duplicates(inplace=True))

  print(df1.dropna(inplace=True))
  print(df2.dropna(inplace=True))
  print(df3.dropna(inplace=True))
  print(df4.dropna(inplace=True))
  return df1,df2,df3,df4
transform(df1,df2,df3,df4)
```

```python
def load(df1,df2,df3,df4):
  df1.to_csv("myvideos.csv",index=False)
  df2.to_csv("mychannels.csv",index=False)
  df3.to_csv("myvideometrics.csv",index=False)
  df4.to_csv("mytraffics.csv",index=False)
  return df1,df2,df3,df4



load(df1,df2,df3,df4)
```
